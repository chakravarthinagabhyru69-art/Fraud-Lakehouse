```sql
SELECT * FROM workspace.fraudlake.txn_practice
```

```sql
SELECT * FROM workspace.fraudlake.txn_practice 
WHERE amount > 100
```

```sql
SELECT * FROM workspace.fraudlake.txn_practice
WHERE  customer = 'maria'
ORDER BY txn_date DESC
```

```sql
SELECT customer, SUM(amount) AS total_spent
FROM workspace.fraudlake.txn_practice
GROUP BY customer
```

```sql
SELECT customer, merchant, amount, txn_date, COUNT(*) AS copies
FROM workspace.fraudlake.txn_practice
GROUP BY customer, merchant, amount, txn_date
HAVING COUNT(*) > 1
```

```sql

SELECT *,
         ROW_NUMBER() OVER (PARTITION BY customer ORDER BY amount DESC) AS rn
  FROM workspace.fraudlake.txn_practice
```

```sql
SELECT *
FROM (
  SELECT *,
         ROW_NUMBER() OVER (PARTITION BY customer ORDER BY amount DESC) AS rn
  FROM workspace.fraudlake.txn_practice
)
WHERE rn = 1
```

```sql
CREATE OR REPLACE TABLE workspace.fraudlake.customers(
    customer STRING,
    home_city STRING,
    risk_tier STRING,
    opened DATE
);
INSERT INTO workspace.fraudlake.customers VALUES
('maria','chicago', 'low',  DATE'2021-03-15'),
('james', 'Huston', 'medium',  DATE'2019-11-02'),
('aisha', 'seattle', 'low', DATE'2022-07-20');
```

```sql
SELECT t.customer, t.merchant, t.amount, c.home_city, c.risk_tier
FROM workspace.fraudlake.txn_practice AS t
JOIN workspace.fraudlake.customers  AS c 
ON t. customer = c.customer;
```

```sql
-- A: plain JOIN
SELECT c.customer, t.merchant
FROM workspace.fraudlake.customers c
JOIN workspace.fraudlake.txn_practice t ON c.customer = t.customer;

```

```sql

-- B: LEFT JOIN
SELECT c.customer, t.merchant
FROM workspace.fraudlake.customers c
LEFT JOIN workspace.fraudlake.txn_practice t ON c.customer = t.customer;
```
