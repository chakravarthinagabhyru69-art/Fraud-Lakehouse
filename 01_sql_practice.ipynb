```sql
SELECT * FROM workspace.fraudlake.txn_practice
```

```sql
SELECT * FROM workspace.fraudlake.txn_practice 
WHERE amount > 100
```

```sql
SELECT * FROM workspace.fraudlake.txn_practice
WHERE  customer = 'maria'
ORDER BY txn_date DESC
```

```sql
SELECT customer, SUM(amount) AS total_spent
FROM workspace.fraudlake.txn_practice
GROUP BY customer
```

```sql
SELECT customer, merchant, amount, txn_date, COUNT(*) AS copies
FROM workspace.fraudlake.txn_practice
GROUP BY customer, merchant, amount, txn_date
HAVING COUNT(*) > 1
```

```sql

SELECT *,
         ROW_NUMBER() OVER (PARTITION BY customer ORDER BY amount DESC) AS rn
  FROM workspace.fraudlake.txn_practice
```

```sql
SELECT *
FROM (
  SELECT *,
         ROW_NUMBER() OVER (PARTITION BY customer ORDER BY amount DESC) AS rn
  FROM workspace.fraudlake.txn_practice
)
WHERE rn = 1
```
